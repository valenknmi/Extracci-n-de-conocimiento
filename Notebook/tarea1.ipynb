```python
import pandas as pd
```

```python
import numpy as np
```

```python
import random
```

```python
from datetime import datetime, timedelta
```

```python
random.seed(42)
```

```python
np.random.seed(42)
```

```python
n_publicaciones = 1000000
```

```python
influencers = [
    "ana_viaja",
    "cancun_life",
    "playas_mex",
    "travel_life",
    "vida_en_mar",
    "turismo_cancun",
    "vibras_nuevas",
    "band-life"
]

tipos_post = [
    "foto",
    "reel",
    "historia",
    "video"
]
```

```python
dataset = pd.DataFrame({
    
    "influencer": [
        random.choice(influencers)
        for _ in range(n_publicaciones)
    ],
    
    "likes": [
        random.randint(100, 50000)
        for _ in range(n_publicaciones)
    ],
    
    "comentarios": [
        random.randint(0, 5000)
        for _ in range(n_publicaciones)
    ],
    
    "fecha": [
        datetime(2025,1,1) + timedelta(days=random.randint(0,365))
        for _ in range(n_publicaciones)
    ],
    
    "tipo_post": [
        random.choice(tipos_post)
        for _ in range(n_publicaciones)
    ]
})
```
